## SQL Practice Question
Working with county-level data containing:
- county_name
- state
- unemployment_rate
### Create a report showing:
- county_name
- state
- unemployment_rate
- a new column called `employment_risk`
### Classification rules:
- unemployment_rate less than 4% → Low Risk
- unemployment_rate between 4% and 7% → Moderate Risk
- unemployment_rate between 7% and 10% → High Risk
- unemployment_rate greater than 10% → Critical Risk
- NULL unemployment_rate - Data Missing

### Only include counties from:
- West Virginia
- Virginia
- Maryland

Return all counties sorted from highest unemployment rate to lowest.
```

````sql
SELECT *
FROM workspace.appalachia.silver_county_dim
WHERE state_name IN ('West Virginia', 'Virginia', 'Maryland')
````
